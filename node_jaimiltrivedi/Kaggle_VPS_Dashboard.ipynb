# 🚀 Kaggle Personal VPS Controller (Master Cockpit)

This notebook converts your Kaggle GPU session into a full-featured personal VPS accessible from any web browser.

### ⚠️ Essential Kaggle Settings (Right Sidebar):
1. **Internet**: Turn **ON** (Required for Cloudflare tunnel & packages)
2. **Accelerator**: Select **GPU T4 x2** (or GPU P100 / CPU)

### ⚡ What this gives you in a single web dashboard:
- **Live System Telemetry**: Real-time CPU, RAM, Disk, and Dual T4 GPU monitoring.
- **Multi-Tab Web Terminal**: Full ANSI bash terminal (Xterm.js) with root privileges.
- **Web File Manager**: Browse `/kaggle/working`, upload, download, and edit files in-browser.
- **1-Click App Store**: VS Code Server, JupyterLab, Ollama, Streamlit.
- **Package Manager**: Install Python PIP and Linux APT packages directly from UI.
- **Port Tunneling**: 1-click Cloudflare tunnels for exposing custom web apps.
- **Keep-Alive**: Built-in heartbeat to prevent browser tabs from sleeping.

```python
# ==============================================================================
# 🚀 KAGGLE VPS 1-CLICK LAUNCHER (REAL-TIME STREAMING)
# Paste this entire code into a Kaggle Notebook cell and click "Run Cell".
# Requirements in Kaggle right-sidebar Settings:
#   - Accelerator: GPU T4 x2 (or GPU P100 / CPU)
#   - Internet: Internet On (Turn ON!)
# ==============================================================================

import os
import sys
import base64
import subprocess

print("=" * 65, flush=True)
print("  🚀 INITIALIZING YOUR PERSONAL KAGGLE VPS...", flush=True)
print("=" * 65, flush=True)

# 1. Install prerequisites
print("[1/3] Installing required packages (FastAPI, Uvicorn, psutil)...", flush=True)
subprocess.check_call([
    sys.executable, "-m", "pip", "install", "-q",
    "fastapi", "uvicorn", "psutil", "python-multipart", "websockets"
])

# 2. Extract VPS Controller Server
print("[2/3] Extracting VPS Controller and Web Interface...", flush=True)
SERVER_B64 = ""
server_path = "/tmp/kaggle_vps_server.py"
with open(server_path, "wb") as f:
    f.write(base64.b64decode(SERVER_B64))

# 3. Launch Server & Stream Output in Real-Time
print("[3/3] Starting VPS Controller daemon with Cloudflare tunnel...", flush=True)
proc = subprocess.Popen(
    [sys.executable, "-u", server_path],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)

displayed_button = False
for line in iter(proc.stdout.readline, ""):
    print(line, end="", flush=True)
    if not displayed_button and "trycloudflare.com" in line and "https://" in line:
        for word in line.split():
            if "https://" in word and "trycloudflare.com" in word:
                clean_url = word.strip().strip("|").strip()
                displayed_button = True
                try:
                    from IPython.display import display, HTML
                    display(HTML(f"""
                    <div style="margin: 20px 0; padding: 20px; background: linear-gradient(135deg, #0f172a, #1e1b4b); border: 2px solid #6366f1; border-radius: 12px; text-align: center; box-shadow: 0 10px 25px -5px rgba(99, 102, 241, 0.3);">
                        <h2 style="color: #ffffff; margin: 0 0 10px 0; font-family: ui-sans-serif, system-ui, sans-serif;">🎉 Your Kaggle VPS Cockpit is Live!</h2>
                        <p style="color: #94a3b8; margin: 0 0 16px 0; font-family: ui-sans-serif, system-ui, sans-serif; font-size: 14px;">Click the button below to open your full VPS dashboard in a new browser tab:</p>
                        <a href="{clean_url}" target="_blank" style="display: inline-block; background: #4f46e5; color: #ffffff; padding: 14px 32px; font-weight: bold; font-size: 16px; border-radius: 8px; text-decoration: none; box-shadow: 0 4px 14px rgba(79, 70, 229, 0.5);">
                            👉 Open VPS Cockpit Dashboard
                        </a>
                        <p style="color: #38bdf8; margin: 12px 0 0 0; font-family: monospace; font-size: 13px;">Direct Link: <a href="{clean_url}" target="_blank" style="color: #38bdf8;">{clean_url}</a></p>
                    </div>
                    """))
                except Exception:
                    pass
```
