```python
# 1. Install prerequisites & Ollama
!apt-get update -qq && apt-get install -y -qq zstd curl >/dev/null 2>&1
!curl -fsSL https://ollama.com/install.sh | sh

# 2. Download Cloudflared
!wget -q -O /tmp/cloudflared https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64
!chmod +x /tmp/cloudflared

# 3. Start Ollama on Custom Port 8008 & Load LLaMA 3.2
import os, subprocess, time
os.environ['OLLAMA_HOST'] = '0.0.0.0:8008'
subprocess.Popen(['ollama', 'serve'], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(3)
!ollama pull llama3.2:1b

# 4. Expose LLaMA 3.2 on secure Cloudflare tunnel
print('\n' + '='*65)
print('  🎉 GOOGLE COLAB TESLA T4 GPU IS LIVE ON PORT 8008!')
print('='*65 + '\n')
!/tmp/cloudflared tunnel --url http://127.0.0.1:8008
```
